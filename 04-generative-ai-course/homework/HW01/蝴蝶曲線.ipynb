#HW-1 畫函數圖形
###蝴蝶曲線(Butterfly Curve)
使用**參數式**繪圖：
$$
\begin{cases}
x = sin(t) \cdot [exp(cos(t))-2cos(4t)-sin^5(t/12)] \\
y = cos(t) \cdot [exp(cos(t))-2cos(4t)-sin^5(t/12)]
\end{cases}, \quad t \in [0, 12\pi] \\
$$



*使用生成式AI幫助Markdown語法與圖形美化*

```python
%matplotlib inline
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 建立參數 t
t = np.linspace(0, 12*np.pi, 5000)

# 蝴蝶曲線公式
x = np.sin(t) * (np.exp(np.cos(t)) - 2*np.cos(4*t) - np.sin(t/12)**5)
y = np.cos(t) * (np.exp(np.cos(t)) - 2*np.cos(4*t) - np.sin(t/12)**5)

# 畫圖
plt.figure(figsize=(8, 8))
plt.plot(x, y, color="purple")

# 美化
plt.title("Butterfly Curve", fontsize=16)
plt.axis("equal")   # 保持比例
plt.axis("off")     # 去掉座標軸

plt.show()
```
